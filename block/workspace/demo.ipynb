# Demo: persistent filesystem

This notebook lives in the Jupyter working directory (`/home/jovyan/work`). In this lab, that directory is mounted from a persistent block storage volume.

We will write a small file into the working directory and then confirm it is still there after restarting the Jupyter container (and later, after recreating the VM and re-attaching the block volume).

```python
import datetime
from pathlib import Path

p = Path('persisted.txt')
stamp = datetime.datetime.utcnow().isoformat() + 'Z'
p.write_text('Hello from persistent block storage!\nSaved at: ' + stamp + '\n')
print('Wrote:', p.resolve())
print(p.read_text())
```
